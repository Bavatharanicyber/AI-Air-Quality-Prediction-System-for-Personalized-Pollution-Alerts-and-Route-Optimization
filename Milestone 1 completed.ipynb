```python
import pandas as pd
import os

years = [2019, 2020, 2021, 2022, 2023]

input_folder = "/content/drive/MyDrive/Datasets"
output_folder = "/content/drive/MyDrive/Delhi_Filtered"

os.makedirs(output_folder, exist_ok=True)

for year in years:
    print(f"Processing {year}...")

    file_path = f"{input_folder}/cpcb-air-quality-{year}.parquet"
    df = pd.read_parquet(file_path)

    # Filter only Delhi
    df_delhi = df[df["City"] == "Delhi"]

    # Required columns
    df_delhi = df_delhi[
        [
            "Station Name",
            "Timestamp",
            "PM2.5 (µg/m³)",
            "PM10 (µg/m³)",
            "NO2 (µg/m³)",
            "CO (mg/m³)",
            "Ozone (µg/m³)"
        ]
    ]

    output_path = f"{output_folder}/Delhi_{year}.parquet"
    df_delhi.to_parquet(output_path, index=False)

    print(f"{year} completed.")

print("✅ All files processed successfully!")
```

```python
df = pd.read_parquet(
    "/content/drive/MyDrive/Delhi_Filtered/Delhi_Final.parquet"
)
```

```python
POLLUTANT_COLS = [
    "pm25",
    "pm10",
    "no2",
    "co",
    "o3"
]

LIMITS = {
    "pm25": (0, 1000),
    "pm10": (0, 1500),
    "no2": (0, 700),
    "co": (0, 50),
    "o3": (0, 500),
}
```

```python
print("\n[2A] Cleaning each pollutant...")

for col in POLLUTANT_COLS:

    before_na = int(df[col].isna().sum())

    # Replace 999 with NaN
    df.loc[df[col] == 999, col] = np.nan

    # Remove values outside valid range
    low, high = LIMITS[col]

    df.loc[(df[col] < low) | (df[col] > high), col] = np.nan

    after_na = int(df[col].isna().sum())

    removed = after_na - before_na

    print(
        f"     {col:<5}: removed {removed} bad values "
        f"(now {df[col].isna().mean()*100:.1f}% blank)"
    )
```

```python
import pandas as pd

# Load cleaned dataset
df = pd.read_parquet("/content/drive/MyDrive/Delhi_Filtered/Delhi_Cleaned.parquet")

# Convert timestamp
df["timestamp"] = pd.to_datetime(df["timestamp"])

# Sort
df = df.sort_values(["station", "timestamp"])

# Create complete hourly timeline
start = df["timestamp"].min().floor("H")
end = df["timestamp"].max().ceil("H")

hours = pd.date_range(start=start, end=end, freq="1H")

stations = df["station"].unique()

full_index = pd.MultiIndex.from_product(
    [stations, hours],
    names=["station", "timestamp"]
)

df = (
    df.set_index(["station", "timestamp"])
      .reindex(full_index)
      .reset_index()
)

print("Rows after timeline:", len(df))
```

```python
pollutants = ["pm25", "pm10", "no2", "co", "o3"]

for col in pollutants:
    before = df[col].isna().sum()

    df[col] = (
        df.groupby("station")[col]
          .ffill(limit=3)
    )

    after = df[col].isna().sum()

    print(f"{col:5} : filled {before-after} | still blank {after}")
```

```python
df.to_parquet(
    "/content/drive/MyDrive/Delhi_Filtered/Delhi_Hourly.parquet",
    index=False
)

print("Step 2C Completed")
print("Rows:", len(df))
print("Stations:", df["station"].nunique())
```

```python
import pandas as pd
import numpy as np

# Load hourly dataset
df = pd.read_parquet("/content/drive/MyDrive/Delhi_Filtered/Delhi_Hourly.parquet")

print(df.head())

print(df.columns)
```

```python
AQI_BREAKPOINTS = {
    "pm25": [
        (0,30,0,50),
        (31,60,51,100),
        (61,90,101,200),
        (91,120,201,300),
        (121,250,301,400),
        (251,500,401,500)
    ],

    "pm10": [
        (0,50,0,50),
        (51,100,51,100),
        (101,250,101,200),
        (251,350,201,300),
        (351,430,301,400),
        (431,600,401,500)
    ],

    "no2": [
        (0,40,0,50),
        (41,80,51,100),
        (81,180,101,200),
        (181,280,201,300),
        (281,400,301,400),
        (401,1000,401,500)
    ],

    "co": [
        (0,1,0,50),
        (1.1,2,51,100),
        (2.1,10,101,200),
        (10.1,17,201,300),
        (17.1,34,301,400),
        (34.1,50,401,500)
    ],

    "o3": [
        (0,50,0,50),
        (51,100,51,100),
        (101,168,101,200),
        (169,208,201,300),
        (209,748,301,400),
        (749,1000,401,500)
    ]
}

print("Breakpoints Loaded Successfully")
```

```python
def calculate_subindex(value, pollutant):

    if pd.isna(value):
        return np.nan

    for BP_LO, BP_HI, I_LO, I_HI in AQI_BREAKPOINTS[pollutant]:

        if BP_LO <= value <= BP_HI:

            return ((I_HI - I_LO) / (BP_HI - BP_LO)) * (value - BP_LO) + I_LO

    return np.nan


print(calculate_subindex(338, "pm25"))
```

```python
# Calculate sub-index for each pollutant
df["aqi_pm25"] = df["pm25"].apply(lambda x: calculate_subindex(x, "pm25"))
df["aqi_pm10"] = df["pm10"].apply(lambda x: calculate_subindex(x, "pm10"))
df["aqi_no2"]  = df["no2"].apply(lambda x: calculate_subindex(x, "no2"))
df["aqi_co"]   = df["co"].apply(lambda x: calculate_subindex(x, "co"))
df["aqi_o3"]   = df["o3"].apply(lambda x: calculate_subindex(x, "o3"))

# Final AQI = Maximum of all sub-indices
df["aqi"] = df[
    ["aqi_pm25", "aqi_pm10", "aqi_no2", "aqi_co", "aqi_o3"]
].max(axis=1)

print("AQI Calculated Successfully!")
print(df[["station", "timestamp", "aqi"]].head())
```

```python
print(df["aqi"].describe())

print(df["aqi"].isna().sum())
```

```python
# AQI summary

print("\nRows with AQI :", df["aqi"].notna().sum(), "/", len(df))
print("Unique AQI values :", df["aqi"].nunique())
print("Mean :", round(df["aqi"].mean(),2))
print("Std :", round(df["aqi"].std(),2))
print("Min :", round(df["aqi"].min(),2))
print("Max :", round(df["aqi"].max(),2))
```

```python
df["year"] = df["timestamp"].dt.year

print("\nAQI Mean by Year")

yearly = df.groupby("year")["aqi"].agg(["mean","std"])

print(yearly.round(2))
```
